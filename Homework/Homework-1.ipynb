## Problem Setting

We consider linear basis function (LBF) models of the form

$$
f(x) = \sum_{j=0}^{m} \theta_j h_j(x),
$$

which are linear in the parameters $\theta$ even if the basis functions $h_j(x)$ are nonlinear in $x$.

Let $X_h \in \mathbb{R}^{n \times (m+1)}$ be the data (or design) matrix defined by $(X_h)_{ij} = h_j(x_i)$. To fit a model, we have minimized the sum-of-squared-error loss

$$
L(\theta) = \| y - X_h \theta \|^2.
$$

In gradient-based optimization, we used weight updates

$$
\theta_{k+1} = \theta_k - \alpha \nabla L(\theta_k),
$$

where the learning rate $\alpha$ must be chosen carefully. Poor choices can lead to slow convergence or divergence.

Rather than fixing $\alpha$, we can instead choose the step size that minimizes the loss along the current search direction. This idea is called **line search**. At iteration $k$, let $d_k$ be a chosen direction, and we define the 1D function

$$
\phi(\alpha) = L(\theta_k + \alpha d_k),
$$

which measures the loss along the line through $\theta_k$ in direction $d_k$. We then choose $\alpha_k$ to minimize $\phi(\alpha)$.
